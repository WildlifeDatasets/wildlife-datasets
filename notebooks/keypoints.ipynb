```python
import matplotlib.pyplot as plt
import numpy as np

from wildlife_datasets.datasets import *

root = "/data/wildlife_datasets/data"
dataset_clasess = [
    (ATRW, {}),
    (BrownBearHeads, {"load_keypoints": True}),
    (CHIRP, {"load_keypoints": True}),
    (CTai, {}),
    (CZoo, {}),
    (CzechLynxv2, {}),
    (IPanda50, {}),
    (Melops, {"load_keypoints": True}),
    (RedBeeReID, {}),
    (RotwildID_Faces, {}), 
]
```

```python
datasets = {}
for dataset_class, dataset_kwargs in dataset_clasess:
    name = dataset_class.display_name()
    root_dataset = f"{root}/{name}"
    dataset = dataset_class(root_dataset, img_load="auto", **dataset_kwargs)
    datasets[name] = dataset
```

```python
for name, dataset in datasets.items():
    idx = np.where(~dataset.metadata["keypoints"].isnull())[0]
    print(name)
    display(dataset.metadata["keypoints"][idx[0]])
```

```python
for name, dataset in datasets.items():
    idx = np.where(~dataset.metadata["keypoints"].isnull())[0]
    dataset.plot_keypoints(idx[0])
    plt.title(name)
```
